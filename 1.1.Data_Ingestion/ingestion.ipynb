## Web Based Loaders

```python
from langchain_community.document_loaders import WebBaseLoader
loader=WebBaseLoader(web_paths=('https://en.wikipedia.org/wiki/Speech/',),)
loader.load()
```

### If we want the info from website from specific classes then...

```python
from langchain_community.document_loaders import WebBaseLoader
import bs4
loader=WebBaseLoader(web_paths=('https://en.wikipedia.org/wiki/Speech/',),
                     bs_kwargs=dict(parse_only=bs4.SoupStrainer(
                         class_=('mw-body-header','vector-body-before-content','hatnote')
                     )))
loader.load()
```
