| Splitter Type              | Works On   | Purpose                                                         |
|-----------------------------|------------|-----------------------------------------------------------------|
| **Recursive JSON Splitter** | JSON data  | Splits nested JSON into smaller, meaningful chunks while keeping structure |


```python
import json
import requests # For making HTTP requests

json_data=requests.get("https://api.smith.langchain.com/openai.json").json()
json_data
```

Here they have nested data of json so we will use Recursive Splitter


```python
from langchain_text_splitters import RecursiveJsonSplitter

json_splitter=RecursiveJsonSplitter(max_chunk_size=300)
json_chunks=json_splitter.split_json(json_data)
json_chunks
```
