| Splitter Type            | Works On      | Purpose                                                |
|---------------------------|---------------|--------------------------------------------------------|
| **Recursive Text Splitter** | Any text      | Splits smartly by structure (paragraph → sentence → word) |
| **Text Splitter**           | Plain text    | Basic splitter, divides by size or tokens              |
| **HTML Text Splitter**      | HTML/webpages | Removes tags and extracts readable content             |


```python
from langchain_text_splitters import HTMLHeaderTextSplitter

html_string='''<!DOCTYPE html>
<html>
<head>
  <title>Two Div Example</title>
</head>
<body>
  <div>
    <h1>First Header</h1>
    <h2>First Div</h2>
    <p>This is the content inside the first div.</p>
  </div>

  <div>
    <h2>Second Div</h2>
    <p>This is the content inside the second div.</p>
  </div>
</body>
</html>'''


header_to_split_on=[
    ("h1","Header 1"),
    ("h2","Header 2")
]

html_splitter = HTMLHeaderTextSplitter(header_to_split_on)
html_header_splits = html_splitter.split_text(html_string)
html_header_splits
```

```python
url="https://prasarbharati.gov.in/dd-india-homepage/"

header_to_split_on=[
    ("h1","Header 1"),
    ("h2","Header 2"),
    ("h3","Header 3")
]

html_splitter = HTMLHeaderTextSplitter(header_to_split_on)
html_header_splits = html_splitter.split_text_from_url(url)
html_header_splits
```
