```python
from langchain_community.embeddings import OllamaEmbeddings

embeddings=(
    OllamaEmbeddings(model="gemma:2b")  ##By default it uses the model "llama2"
)
```

```python
embeddings
```

```python
r1=embeddings.embed_documents(
    [
        "Hello world this is a test",
        "LangChain is a framework for developing can be used for chatbotsuestion-Answering (GQA), summarization, and much more.",
    ]
)
```

```python
r1
```

```python
embeddings.embed_query("Hello world this is a test")
```
