Sentence Transformers (a.k.a. sentence-transformers) is a Python library built on top of Hugging Face Transformers.

It is used to convert sentences or paragraphs into embeddings — numerical vectors that represent semantic meaning.

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ['HF_TOKEN']=os.getenv('HF_TOKEN')
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings=HuggingFaceEmbeddings(model="all-MiniLM-L6-v2")
```

```python
text="Hello world this is a test sentence for embeddings"
vector=embeddings.embed_query(text)
vector
```
